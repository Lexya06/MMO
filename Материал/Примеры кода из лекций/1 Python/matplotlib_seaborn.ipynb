## Часть 1: Основы matplotlib

### Пример 1. Линейный график (plt.plot)

Линейный график - это базовый тип визуализации, предназначенный для отображения непрерывных данных и зависимостей между переменными. Идеально подходит для визуализации временных рядов, функциональных зависимостей, трендов и динамики изменения показателей. Используйте линейные графики, когда нужно показать, как значение меняется вдоль непрерывной оси (обычно времени или другой упорядоченной переменной), выявить тренды, сезонность или паттерны в данных.
Построение простого графика функции sin(x) на интервале от 0 до 2π. Это самый базовый тип визуализации в matplotlib, с которого начинается изучение библиотеки.


```python
import numpy as np
import matplotlib.pyplot as plt

# Генерируем данные
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Строим график
plt.figure(figsize=(8, 4))
plt.plot(x, y, color='#20808D', linewidth=2)
plt.title('Функция sin(x)')
plt.xlabel('x (радианы)')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Пример 2. Настройка осей и заголовков

Настройка визуальных элементов графика - важная часть создания информативной и профессионально выглядящей визуализации. Правильное оформление заголовков, подписей осей, легенды и сетки повышает читаемость графика и помогает аудитории быстрее понять представленные данные. Применяйте эти настройки для создания презентационных материалов, научных публикаций и отчётов, где важна ясность и визуальная привлекательность.
Демонстрация настройки внешнего вида графика: заголовок, подписи осей, сетка и легенда.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 50)
y = x ** 2

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'o-', color='#A84B2F', markersize=4, label='y = x²')
plt.xlabel('Ось X', fontsize=12, fontweight='bold')
plt.ylabel('Ось Y', fontsize=12, fontweight='bold')
plt.title('Настройка осей и заголовков', fontsize=14, fontweight='bold', pad=15)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left', fontsize=11, framealpha=0.9)
plt.xlim(0, 10)
plt.ylim(0, 105)
plt.tight_layout()
plt.show()
```

### Пример 3. Несколько линий на одном графике

Многолинейный график - эффективный способ сравнения нескольких рядов данных на одной координатной плоскости. Используется для сопоставления динамики различных показателей, выявления корреляций и различий между группами данных. Применяйте, когда необходимо сравнить временные ряды нескольких продуктов, регионов, категорий или показать одновременное изменение связанных метрик (например, температура и влажность, доходы и расходы).

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)

plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x), color='#20808D', linewidth=2, label='sin(x)')
plt.plot(x, np.cos(x), color='#A84B2F', linewidth=2, linestyle='--', label='cos(x)')
plt.title('Сравнение sin(x) и cos(x)')
plt.xlabel('x (радианы)')
plt.ylabel('Значение')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='grey', linewidth=0.5)
plt.tight_layout()
plt.show()
```

### Пример 4. Столбчатая диаграмма (plt.bar)

Столбчатая диаграмма (bar chart) - классический инструмент для сравнения дискретных категорий или групп данных. Высота столбца пропорциональна значению показателя, что делает различия наглядными и легко интерпретируемыми. Используйте столбчатые диаграммы для сравнения продаж по месяцам, выручки по регионам, количественных показателей по категориям товаров, результатов опросов. Особенно эффективны, когда категорий немного (до 10-12) и важно показать точные значения.
Сравнение продаж по месяцам. Вертикальные столбцы для дискретных категорий.


```python
import matplotlib.pyplot as plt

месяцы = ['Янв', 'Фев', 'Мар', 'Апр', 'Май', 'Июн']
продажи = [120, 135, 148, 162, 155, 170]

plt.figure(figsize=(8, 5))
bars = plt.bar(месяцы, продажи, color='#20808D', edgecolor='white', width=0.6)

for bar, val in zip(bars, продажи):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
             str(val), ha='center', va='bottom', fontsize=10)

plt.title('Продажи по месяцам (тыс. руб.)')
plt.ylabel('Продажи')
plt.ylim(0, 195)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Пример 5. Горизонтальная столбчатая диаграмма (plt.barh)

Горизонтальная столбчатая диаграмма - вариация обычной столбчатой диаграммы, где категории расположены вдоль вертикальной оси, а значения - вдоль горизонтальной. Особенно удобна для рейтингов, топ-списков и ситуаций, когда названия категорий длинные и не помещаются под вертикальными столбцами. Применяйте для отображения рейтингов (топ-10 продуктов, языков программирования, компаний), результатов голосований, сравнения показателей с длинными названиями.
Рейтинг языков программирования. Горизонтальные столбцы удобны для рейтингов.


```python
import matplotlib.pyplot as plt

языки = ['Python', 'JavaScript', 'Java', 'C++', 'C#', 'TypeScript', 'Go', 'Rust']
индекс = [28.5, 18.2, 15.8, 10.3, 7.5, 6.1, 4.8, 3.2]

sorted_data = sorted(zip(индекс, языки))
языки_s = [d[1] for d in sorted_data]
индекс_s = [d[0] for d in sorted_data]

plt.figure(figsize=(8, 5))
colors_list = ['#BCE2E7'] * len(языки_s)
colors_list[-1] = '#20808D'
plt.barh(языки_s, индекс_s, color=colors_list, edgecolor='white', height=0.6)
plt.xlabel('Индекс популярности (%)')
plt.title('Рейтинг языков программирования (2026)')
plt.tight_layout()
plt.show()
```

### Пример 6. Круговая диаграмма (plt.pie)

Круговая диаграмма (pie chart) - визуализация для отображения структуры целого, где каждый сектор представляет долю от 100%. Эффективна для демонстрации процентного соотношения частей в составе целого. Используйте круговые диаграммы для показа структуры расходов или доходов, долей рынка компаний, распределения ресурсов, состава аудитории. Важно: для читаемости ограничивайте количество секторов (не более 5-7), иначе диаграмма становится перегруженной и трудночитаемой.
Структура расходов домохозяйства. Не более 5 категорий для читаемости.


```python
import matplotlib.pyplot as plt

категории = ['Жильё', 'Еда', 'Транспорт', 'Развлечения', 'Накопления']
доли = [35, 25, 15, 10, 15]
цвета = ['#20808D', '#A84B2F', '#1B474D', '#FFC553', '#944454']
explode = (0.05, 0, 0, 0, 0)

plt.figure(figsize=(7, 7))
plt.pie(доли, labels=категории, autopct='%1.0f%%', colors=цвета, explode=explode,
        startangle=90, textprops={'fontsize': 11})
plt.title('Распределение расходов домохозяйства')
plt.tight_layout()
plt.show()
```

### Пример 7. Точечная диаграмма (plt.scatter)

Точечная диаграмма (scatter plot) - мощный инструмент для визуализации взаимосвязи между двумя или более переменными. Каждая точка представляет одно наблюдение с координатами по осям X и Y, а дополнительные измерения можно кодировать через размер и цвет точек. Применяйте для исследования корреляций между переменными, выявления кластеров и выбросов в данных, визуализации результатов машинного обучения (например, кластеризации), анализа многомерных данных (3-4 измерения через координаты, размер и цвет).
Связь между переменными. Цвет и размер точек кодируют дополнительные измерения.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n = 100
x = np.random.rand(n) * 100
y = x * 0.7 + np.random.randn(n) * 15
size = np.random.rand(n) * 200 + 20
color = np.random.rand(n)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(x, y, s=size, c=color, cmap='viridis', alpha=0.7,
                      edgecolors='white', linewidth=0.5)
plt.colorbar(scatter, label='Категория')
plt.xlabel('Переменная X')
plt.ylabel('Переменная Y')
plt.title('Точечная диаграмма с цветовой кодировкой')
plt.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
```

### Пример 8. Гистограмма (plt.hist)

Гистограмма - статистический инструмент для визуализации распределения непрерывных данных. График разбивает диапазон значений на интервалы (bins) и показывает частоту попадания наблюдений в каждый интервал. Используйте гистограммы для оценки формы распределения данных (нормальное, скошенное, бимодальное), выявления выбросов и аномалий, проверки предположений статистических тестов, визуализации распределения возраста, дохода, результатов измерений. Часто дополняется наложением теоретической кривой распределения (KDE).
Распределение нормально распределённых данных с наложенной кривой плотности.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
data = np.random.normal(loc=170, scale=10, size=1000)

plt.figure(figsize=(8, 5))
n, bins, patches = plt.hist(data, bins=30, color='#20808D', edgecolor='white',
                             alpha=0.8, density=True)

xmin, xmax = plt.xlim()
x_line = np.linspace(xmin, xmax, 100)
pdf = stats.norm.pdf(x_line, loc=170, scale=10)
plt.plot(x_line, pdf, color='#A84B2F', linewidth=2, label='Плотность N(170, 10)')

plt.xlabel('Рост (см)')
plt.ylabel('Плотность')
plt.title('Распределение роста (N=1000)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Пример 9. Несколько графиков (subplots)

Сетка подграфиков (subplots) - способ размещения нескольких независимых графиков на одной фигуре в виде сетки. Позволяет компактно представить множество визуализаций и облегчает сравнение между ними. Применяйте для создания дашбордов и сводных отчётов, сравнения различных аспектов одного датасета, отображения данных по разным группам или временным периодам, комбинирования разных типов визуализаций для комплексного анализа.
Сетка 2×2 с разными типами графиков.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(0, 2 * np.pi, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, np.sin(x), color='#20808D')
axes[0, 0].set_title('Линейный')
axes[0, 0].grid(True, alpha=0.3)

cat = ['A', 'B', 'C', 'D', 'E']
vals = [23, 45, 56, 78, 32]
axes[0, 1].bar(cat, vals, color='#A84B2F')
axes[0, 1].set_title('Столбчатая')

axes[1, 0].scatter(np.random.rand(50), np.random.rand(50), c='#1B474D', alpha=0.6, s=60)
axes[1, 0].set_title('Точечная')

axes[1, 1].hist(np.random.randn(500), bins=25, color='#FFC553', edgecolor='white')
axes[1, 1].set_title('Гистограмма')

fig.suptitle('Сетка 2×2 с разными типами графиков', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Пример 10. Сохранение графика (savefig)

Экспорт графика в файл - необходимая функция для включения визуализаций в отчёты, презентации и публикации. Matplotlib поддерживает множество форматов (PNG, PDF, SVG, JPG) с настройкой качества и параметров вывода. Используйте PNG для веб-страниц и презентаций (с высоким DPI для печати), PDF для научных публикаций и документов (векторный формат), SVG для веб-графики с возможностью масштабирования. Параметр dpi=300 обеспечивает качество, достаточное для печати.
Экспорт в PNG и PDF с настройкой качества.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.exp(-x/3) * np.sin(2 * x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, color='#20808D', linewidth=2)
ax.fill_between(x, y, alpha=0.1, color='#20808D')
ax.set_title('Затухающие колебания')
ax.set_xlabel('Время (с)')
ax.set_ylabel('Амплитуда')
ax.grid(True, alpha=0.3)

# Сохранение в разных форматах
fig.savefig('график.png', dpi=300, bbox_inches='tight', facecolor='white')
fig.savefig('график.pdf', bbox_inches='tight', facecolor='white')
print('Графики сохранены: график.png, график.pdf')
plt.show()
```

### Пример 11. Аннотации и стрелки (annotate)

Аннотации с стрелками - инструмент для выделения ключевых точек, событий или аномалий на графике с поясняющими комментариями. Стрелка привлекает внимание к важному элементу, а текстовое пояснение даёт контекст. Используйте для выделения экстремумов (максимумов, минимумов), важных дат и событий на временных рядах (запуск продукта, кризис), аномалий и выбросов в данных, ключевых точек перегиба или пересечения. Особенно полезно в презентациях и отчётах для руководства.
Выделение ключевых точек на графике.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, color='#20808D', linewidth=2)

x_max = x[np.argmax(y)]
x_min = x[np.argmin(y)]

ax.annotate('Максимум\n(π/2, 1.0)',
    xy=(x_max, 1.0), xytext=(x_max + 1, 0.7), fontsize=10,
    arrowprops=dict(arrowstyle='->', color='#A84B2F', lw=1.5),
    bbox=dict(boxstyle='round,pad=0.3', facecolor='#FFC553', alpha=0.8))

ax.annotate('Минимум\n(3π/2, -1.0)',
    xy=(x_min, -1.0), xytext=(x_min - 1.5, -0.5), fontsize=10,
    arrowprops=dict(arrowstyle='->', color='#A84B2F', lw=1.5),
    bbox=dict(boxstyle='round,pad=0.3', facecolor='#FFC553', alpha=0.8))

ax.set_title('Аннотации на графике sin(x)')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Пример 12. Заливка областей (fill_between)

Заливка областей между кривыми - способ визуализации диапазона неопределённости, доверительных интервалов или разницы между двумя функциями. Заполненная область привлекает внимание к зоне вариативности данных. Применяйте для отображения доверительных интервалов прогнозов и моделей, диапазонов погрешности измерений, зон между верхней и нижней границей показателя, областей между двумя сценариями (оптимистичный/пессимистичный). Прозрачность заливки позволяет видеть основную линию тренда.
Доверительный интервал вокруг основной линии данных.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(0, 10, 50)
y = 2 * x + 1
y_err = np.random.uniform(1, 3, size=len(x))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, color='#20808D', linewidth=2, label='Среднее')
ax.fill_between(x, y - y_err, y + y_err, color='#20808D', alpha=0.15,
                label='Доверительный интервал')

y_noisy = y + np.random.randn(len(x)) * 2
ax.scatter(x, y_noisy, c='#A84B2F', s=15, alpha=0.6, label='Наблюдения', zorder=5)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Доверительный интервал')
ax.legend()
ax.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
```

### Пример 13. Двойная ось Y (twinx)

График с двумя осями Y - решение для одновременного отображения двух показателей с разными масштабами и единицами измерения на одном графике. Левая и правая оси Y независимы, что позволяет корректно визуализировать данные разных порядков. Используйте для сопоставления показателей с разными единицами (температура и осадки, цена и объём продаж, доходы и рентабельность), выявления связей между разнородными метриками, экономии места при представлении связанных данных. Важно: цветовое кодирование осей и линий должно совпадать для лучшей читаемости.
Температура и осадки на одном графике с двумя осями Y.


```python
import matplotlib.pyplot as plt

месяцы = ['Янв','Фев','Мар','Апр','Май','Июн','Июл','Авг','Сен','Окт','Ноя','Дек']
температура = [-8, -6, 0, 8, 16, 20, 22, 20, 14, 7, 0, -5]
осадки = [40, 35, 38, 45, 55, 70, 80, 75, 60, 50, 45, 42]

fig, ax1 = plt.subplots(figsize=(10, 5))

line1 = ax1.plot(месяцы, температура, 'o-', color='#A84B2F', linewidth=2, label='Температура')
ax1.set_ylabel('Температура (°C)', color='#A84B2F')
ax1.tick_params(axis='y', labelcolor='#A84B2F')

ax2 = ax1.twinx()
line2 = ax2.bar(месяцы, осадки, alpha=0.3, color='#20808D', label='Осадки')
ax2.set_ylabel('Осадки (мм)', color='#20808D')
ax2.tick_params(axis='y', labelcolor='#20808D')

ax1.set_title('Климатограмма Минска')
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
plt.tight_layout()
plt.show()
```

### Пример 14. Настройка стиля (plt.style.use)

Встроенные стили matplotlib - предустановленные наборы параметров оформления графиков, позволяющие быстро изменить общий вид визуализаций. Стили влияют на цветовые схемы, фоны, сетки, шрифты и другие элементы. Используйте для быстрого переключения между профессиональными темами оформления, адаптации графиков под корпоративный стиль или требования публикации, создания тёмных тем для презентаций, экспериментов с визуальным стилем. Популярные стили: 'ggplot', 'seaborn-v0_8-*', 'bmh', 'dark_background'.
Встроенные стили matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)

стили = ['default', 'seaborn-v0_8-whitegrid', 'ggplot', 'dark_background']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, style_name in zip(axes.flat, стили):
    with plt.style.context(style_name):
        ax.plot(x, y1, label='sin(x)')
        ax.plot(x, y2, label='cos(x)')
        ax.set_title(f'Стиль: {style_name}', fontsize=10)
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)

fig.suptitle('Сравнение стилей matplotlib', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Список всех доступных стилей:
# print(plt.style.available)
```

### Пример 15. Гистограмма с KDE (sns.histplot)

Гистограмма с оценкой плотности ядра (KDE) - комбинация классической гистограммы со сглаженной кривой плотности вероятности. KDE (Kernel Density Estimation) даёт более гладкое представление о форме распределения, чем дискретные столбцы гистограммы. Используйте для первичного исследования распределения данных (EDA), выявления мультимодальности (нескольких пиков) в распределении, визуализации скошенности и асимметрии данных, сравнения теоретического и эмпирического распределений. KDE помогает увидеть общую форму распределения независимо от выбора количества bins.
Гистограмма с оценкой плотности ядра.


```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.concatenate([
    np.random.normal(60, 10, 500),
    np.random.normal(85, 8, 300)
])

plt.figure(figsize=(8, 5))
sns.histplot(data, bins=40, kde=True, color='#20808D', edgecolor='white',
             stat='density', alpha=0.7)
plt.xlabel('Значение')
plt.ylabel('Плотность')
plt.title('Бимодальное распределение с KDE')
plt.tight_layout()
plt.show()
```

### Пример 16. Ящик с усами (sns.boxplot)

Диаграмма "ящик с усами" (boxplot) - компактный способ отображения статистического распределения данных через квартили. Показывает медиану (линия в центре ящика), первый и третий квартили (границы ящика), выбросы (точки за пределами усов). Применяйте для сравнения распределений между группами или категориями, быстрой оценки центральной тенденции и разброса данных, выявления выбросов и асимметрии в данных, визуализации результатов A/B тестирования. Особенно эффективен при сравнении нескольких групп одновременно.
Boxplot - медиана, квартили и выбросы.


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
df = pd.DataFrame({
    'Предмет': np.repeat(['Математика', 'Физика', 'Информатика', 'Химия'], 50),
    'Балл': np.concatenate([
        np.random.normal(72, 12, 50),
        np.random.normal(68, 15, 50),
        np.random.normal(80, 10, 50),
        np.random.normal(65, 18, 50)
    ])
})

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Предмет', y='Балл',
            palette=['#20808D', '#A84B2F', '#1B474D', '#FFC553'],
            linewidth=1.5,
            flierprops=dict(marker='o', markersize=4))
plt.title('Распределение баллов по предметам')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Пример 17. Скрипичная диаграмма (sns.violinplot)

Скрипичная диаграмма (violin plot) - гибрид boxplot и KDE-графика, показывающий как статистические характеристики (квартили), так и форму распределения данных. Ширина "скрипки" в каждой точке пропорциональна плотности данных на этом уровне. Используйте когда важно видеть не только медиану и квартили, но и полную форму распределения, для обнаружения мультимодальности в группах данных, сравнения сложных распределений между категориями, визуализации данных с нестандартным распределением. Особенно ценна при наличии бимодальных или асимметричных распределений.
Объединяет boxplot и KDE. Полезно при мультимодальных данных.


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
df = pd.DataFrame({
    'Группа': np.repeat(['A', 'B', 'C'], 100),
    'Значение': np.concatenate([
        np.random.normal(50, 10, 100),
        np.concatenate([np.random.normal(40, 5, 50), np.random.normal(65, 5, 50)]),
        np.random.exponential(15, 100) + 30
    ])
})

plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Группа', y='Значение',
    palette=['#20808D', '#A84B2F', '#1B474D'],
    inner='quartile', linewidth=1.2, saturation=0.85)
plt.title('Скрипичная диаграмма трёх групп')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Пример 18. Тепловая карта (sns.heatmap)

Тепловая карта (heatmap) - матричная визуализация, где значения кодируются цветом. Интенсивность или оттенок цвета соответствует величине показателя в ячейке. Особенно эффективна для больших таблиц данных. Применяйте для визуализации корреляционных матриц (выявление взаимосвязей между переменными), отображения матриц путаницы (confusion matrix) в машинном обучении, показа двумерных распределений или таблиц сопряжённости, временных паттернов (день недели × час суток). Аннотации значениями в ячейках повышают информативность.
Корреляционная матрица с аннотациями.


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
n = 200
data = {
    'Рост': np.random.normal(175, 10, n),
    'Вес': np.random.normal(75, 12, n),
    'Возраст': np.random.normal(35, 10, n),
    'Доход': np.random.normal(50000, 15000, n),
    'Стаж': np.random.normal(10, 5, n),
}
data['Вес'] = data['Рост'] * 0.5 + np.random.randn(n)*8
data['Доход'] = data['Стаж'] * 3000 + np.random.randn(n)*10000

df = pd.DataFrame(data)
corr = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r',
    center=0, vmin=-1, vmax=1,
    square=True, linewidths=0.5,
    cbar_kws={'shrink': 0.8, 'label': 'Корреляция'})
plt.title('Корреляционная матрица')
plt.tight_layout()
plt.show()
```

### Пример 19. Парный график (sns.pairplot)

Парный график (pairplot) - матрица графиков, показывающая попарные зависимости между всеми числовыми переменными в датасете. На диагонали обычно располагаются распределения отдельных переменных (гистограммы или KDE), вне диагонали - scatter plots для пар переменных. Используйте для комплексного исследовательского анализа данных (EDA), быстрого обзора всех взаимосвязей в многомерном датасете, выявления кластеров и сепарабельности классов, проверки предположений перед построением моделей. Параметр hue позволяет раскрасить точки по категориям для анализа разделимости классов.
Матрица попарных зависимостей для датасета iris.


```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')

g = sns.pairplot(iris, hue='species',
    palette=['#20808D', '#A84B2F', '#1B474D'],
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 's': 30, 'edgecolor': 'white'},
    height=2.2, aspect=1)

g.fig.suptitle('Парный график (iris)', y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Пример 20. Категориальный график (sns.catplot)

Категориальный график (catplot) - универсальная функция seaborn для работы с категориальными данными. Поддерживает различные типы графиков (bar, box, violin, strip, swarm) и автоматическую группировку по дополнительным категориям через параметр hue. Применяйте для сравнения групп по категориальным признакам с разбивкой на подгруппы, визуализации результатов экспериментов с несколькими факторами, анализа продаж/метрик по регионам и временным периодам, создания группированных столбчатых диаграмм с планками погрешностей.
Категориальный график с группировкой.


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
df = pd.DataFrame({
    'Город': np.repeat(['Москва', 'Минск', 'Киев', 'Астана'], 60),
    'Квартал': np.tile(np.repeat(['Q1', 'Q2', 'Q3'], 20), 4),
    'Продажи': np.random.normal(100, 20, 240) + np.repeat([30, 10, 15, 5], 60)
})

g = sns.catplot(data=df, x='Город', y='Продажи',
    hue='Квартал', kind='bar',
    palette=['#20808D', '#A84B2F', '#FFC553'],
    height=5, aspect=1.5,
    errorbar='sd', alpha=0.85)

g.set_axis_labels('Город', 'Продажи (тыс.)')
g.fig.suptitle('Продажи по городам и кварталам', y=1.02, fontsize=13)
plt.tight_layout()
plt.show()
```

### Пример 21. Регрессионный график (sns.lmplot)

Регрессионный график (lmplot) - scatter plot с наложенной линией линейной регрессии и доверительным интервалом. Автоматически вычисляет и визуализирует линейную зависимость между переменными. Используйте для визуализации и проверки линейных зависимостей между переменными, оценки силы и направления связи (положительная/отрицательная корреляция), сравнения регрессий для разных групп (через параметр hue), презентации результатов регрессионного анализа. Доверительный интервал (по умолчанию 95%) показывает неопределённость оценки.
Линейная регрессия с доверительным интервалом.


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
n = 150
опыт = np.random.uniform(1, 20, n)
зарплата = 30000 + опыт * 5000 + np.random.randn(n)*8000
группа = np.where(опыт < 10, 'Junior/Middle', 'Senior')

df = pd.DataFrame({
    'Опыт (лет)': опыт,
    'Зарплата': зарплата,
    'Уровень': группа
})

g = sns.lmplot(data=df, x='Опыт (лет)', y='Зарплата',
    hue='Уровень',
    palette=['#20808D', '#A84B2F'],
    height=5, aspect=1.4,
    scatter_kws={'alpha': 0.5, 's': 30, 'edgecolor': 'white'},
    ci=95)

g.set_axis_labels('Опыт работы (лет)', 'Зарплата (руб./мес.)')
g.fig.suptitle('Регрессия: опыт → зарплата', y=1.02, fontsize=13)
plt.tight_layout()
plt.show()
```

### Пример 22. Настройка стиля seaborn (set_theme)

Настройка стиля и контекста seaborn - система управления визуальным оформлением графиков через предустановленные темы. Параметр style задаёт общий вид (фон, сетка), context - масштаб элементов для разных сценариев использования, palette - цветовую схему. Используйте контекст 'paper' для научных публикаций (мелкие шрифты), 'notebook' для Jupyter notebooks (средний размер), 'talk' для презентаций (крупные элементы), 'poster' для постеров (максимальный размер). Функция set_theme() применяет настройки глобально, sns.reset_defaults() возвращает к стандартным.
Палитры и контексты seaborn.


```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(0, 10, 100)

контексты = ['paper', 'notebook', 'talk', 'poster']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, ctx in zip(axes.flat, контексты):
    sns.set_theme(style='whitegrid', context=ctx, palette='deep')
    ax.plot(x, np.sin(x), label='sin(x)')
    ax.plot(x, np.cos(x), label='cos(x)')
    ax.set_title(f'Контекст: {ctx}')
    ax.legend()

fig.suptitle('Контексты seaborn: paper → poster', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Сброс настроек по умолчанию
sns.reset_defaults()
```
