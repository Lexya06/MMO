# **Именные функции**

В Python функция - это объект. Он принимает аргументы, выполняет с ними определённые операции и возвращает результат (значение). Функция определяется с помощью инструкции def, после которой следует имя функции.
Инструкция return указывает, что функция должна вернуть.

```python
def greet(name):
    return f"Привет, {name}!"

message = greet("Ольга")
print(message)
```

В примере представлена простейшая функция, которая принимает один параметр. В зависимости от типов данных параметров, результатом функции может быть число или строка.

Структура функции определяется спецификой задачи. Оператор return не используется, если функция выполняет некоторые действия, но не возвращает значения. В этом случае возвращаемое значение — None.

```python
def my_func():
    pass

print(my_func())
```

Здесь в теле функции реализован оператор-заглушка. Его использование равноценно отсутствию операции. Pass может применяться в тех случаях, когда код на текущий момент не написан.

***Оператор return***

***return со значением***

Если при выполнении логики функции интерпретатор Python встречает оператор return, то забирает значение, определённое после оператора, и выполняет выход из функции.


```python
def square_circle(user_radius):
  square = 2 * 3.14 * user_radius
  print(square)

user_radius = float(input('Enter radius '))
result = square_circle(user_radius)
```

Здесь в главную ветку из функции возвращается значение локальной переменной square, не сама переменная, а её значение.




***return без значения***

В функции можно реализовать несколько операторов return. Но по итогам работы функций может быть выполнен только один оператор return, его поток выполнения программы достигнет первым.


```python
def s_calc():
    try:
        r_val = float(input("Укажите радиус: "))
        h_val = float(input("Укажите высоту: "))
    except ValueError:
        return
    s_side = 2 * 3.14 * r_val * h_val
    s_circle = 3.14 * r_val ** 2
    s_full = s_side + 2 * s_circle
    return s_full

print(s_calc())
```

В этом примере предусмотрена ситуация, когда пользователь вводит некорректные данные. Например, вместо числа (радиус, высота) он ввёл строку. При этом реализована обработка исключения, которое возникнет при попытке выполнения арифметической операции умножения со строками. В ветке except при возникновении исключения осуществляется выход из функции без вычисления площади цилиндра.

В результате функция возвращает объект типа None. Такое значение оператор return возвращает по умолчанию, но можно указать его явно: return None.

Важно! Если в функции отсутствует оператор return, не значит, что она ничего не возвращает. На самом деле возвращает. Только это значение не присваивается какой-либо переменной и не выводится на экран. В Python любая функция что-то возвращает. Если оператор return отсутствует, то возвращаемое значение — None.


***Возврат набора значений***

В Python возможно использование оператора return, возвращающего из функции несколько объектов. Достаточно указать их через запятую после оператора return.

```python
def s_calc():
    try:
        r_val = float(input("Укажите радиус: "))
        h_val = float(input("Укажите высоту: "))
    except ValueError:
        return
    s_side = 2 * 3.14 * r_val * h_val
    s_circle = 3.14 * r_val ** 2
    s_full = s_side + 2 * s_circle
    return s_side, s_full

s_side_val, s_full_val = s_calc()
print(f"Площадь боковой пов-ти - {s_side_val}; Полная площадь - {s_full_val}")
```

Функция s_calc() возвращает два значения, присваиваемые переменным s_side_val и s_full_val. Подобное групповое присвоение — важная характеристика Python.

Смысл в том, что перечисление значений через запятую формирует объект типа кортеж (tuple). Присваивая кортеж сразу набору переменных, его элементы сопоставляются переменным. Происходит своего рода распаковка.

То есть, когда функция возвращает набор объектов, на деле она возвращает объект-кортеж с этими объектами. Они упаковываются в кортеж перед возвратом. Если за оператором return следует только одна переменная, её тип сохраняется в исходном состоянии.


***Аргументы функций***

Функция может принимать любое количество параметров или не принимать их вообще. Параметры могут быть позиционные и именованные, обязательные и необязательные.

```python
# позиционные параметры
def func_1(a, b, c):
    return a + b + c

print(func_1(10, 20, 30))

# именованные параметры
def second_func(a, b, c):
    print(f"a - {a}; b - {b}; c - {c}")

second_func(b=10, a=20, c=30)
```

```python
# обязательные параметры
def func_1(a, b, c):
    return a + b + c

print(func_1(10, 20, 30))

# b и c - необязательные параметры
def func_2(a, b=20, c=30):
    return a + b + c

print(func_2(10))
```

```python
def calculate_price(price, quantity, tax=0.2, discount=0):
    total = price * quantity
    total_with_tax = total * (1 + tax)
    final_price = total_with_tax - discount
    return final_price

# Вызов с позиционными (100, 2) и именованным (discount=50) аргументами
print(calculate_price(100, 2, discount=50))
```

В функции calculate_price цена и количество товара передаются как позиционные аргументы, так как они первичны. Процент скидки и налога указываются через именованные параметры, что дает возможность менять их местами или использовать значения по умолчанию. Такой подход делает вызов функции calculate_price(100, 5, discount=10) наглядным: сразу ясно, что число 10 относится именно к скидке, а не к какой-то другой переменной.

Функция может принимать неопределённое число позиционных параметров. В этом случае при описании функции используется конструкция *args.

```python
def my_func(*numbers):
    return numbers

print(my_func(10, "text_1", 20, "text_2"))
```

# ***Анонимные функции (lambda)***

Это функции, содержащие только одно выражение, но выполняющиеся быстрее именных функций. При этом используется оператор lambda. При использовании lambda-функций их необязательно присваивать некоторой переменной, как в случае с именными функциями. lambda-функции (анонимная или несвязанная), в отличие от именных, не требуют оператора return, в остальном — идентичны именным.

```python
square = lambda x: x ** 2
print(square(5))
```

Рассмотрим список студентов, где каждый представлен словарем с именем и оценкой. Если мы хотим отсортировать их по баллу в порядке убывания, мы передаем в метод sort параметр key, значением которого будет лямбда. Она принимает объект студента и возвращает его оценку, что служит инструкцией для алгоритма сортировки. В итоге мы получаем список, выстроенный от самого высокого результата к самому низкому.

```python
students = [
    {"name": "Анна", "score": 88},
    {"name": "Иван", "score": 95},
    {"name": "Олег", "score": 82}
]

students.sort(key=lambda student: student["score"], reverse=True)
print(students)
```

# ***Области видимости переменных в функциях***
Понятие «Область видимости» определяет, когда и в каких точках программы разработчик может применять свои пользовательские объекты (переменные, функции). В Python доступны следующие области видимости: локальная, глобальная, нелокальная.

```python
def calculate_sum(a, b):
    result = a + b   # локальная переменная
    print("Внутри функции:", result)

calculate_sum(3, 5)
# print(result)  # NameError: result is not defined
```

Здесь result существует только внутри calculate_sum, поэтому после завершения функции обратиться к ней уже нельзя. Это удобно для временных промежуточных данных, которые не должны быть доступны во всей программе.

Глобальная область видимости охватывает имена, объявленные на верхнем уровне модуля, то есть вне функций. Такие переменные можно читать из функций, если внутри функции не создается локальная переменная с тем же именем.

```python
message = "Привет, мир"   # глобальная переменная

def show_message():
    print(message)

show_message()
print(message)
```

В этом примере message объявлена вне функции, поэтому она видна и внутри show_message, и в основном коде программы. Если внутри функции присвоить имя без global, Python создаст новую локальную переменную, а не изменит глобальную

```python
counter = 0

def increment():
    global counter
    counter += 1

increment()
print(counter)   # 1
```

Ключевое слово global явно говорит Python, что внутри функции нужно работать именно с глобальной переменной. Без него попытка изменить counter внутри функции привела бы к ошибке или созданию локального имени в зависимости от кода

Нелокальная область видимости появляется во вложенных функциях и означает, что переменная находится не в локальной области текущей функции и не в глобальной, а во внешней охватывающей функции. Для изменения такой переменной используют ключевое слово nonlocal.

```python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20
        print("Внутри inner:", x)

    inner()
    print("Внутри outer после inner:", x)

outer()
```

Здесь x создана в outer, а функция inner меняет именно эту переменную через nonlocal. Если убрать nonlocal, то присваивание x = 20 внутри inner создаст уже локальную переменную x для inner, а не изменит x из outer.

# ***Генераторы списков***

```python
my_list = [2, 4, 6]
new_list = [el+10 for el in my_list]
print(f"Исходный список: {my_list}")
print(f"Новый список: {new_list}")
```

В приведённом примере функцию генератора выполняет выражение: el+10 for el in my_list, где my_list — итерируемый объект. Из него в цикле for поочерёдно извлекаются элементы. Перед инструкцией for указывается действие, которое выполняется над элементом перед добавлением его в новый список.

```python
my_list = [2, 4, 6]
print(f"Исходный список: {my_list}")
new_list = []
for el in my_list:
   new_list.append(el + 10)
print(f"Новый список: {new_list}")
```

```python
#В генератор допускается добавление условия.
my_list = [10, 25, 30, 45, 50]
print(my_list)
new_list = [el for el in my_list if el % 2 == 0]
print(new_list)
```

```python
#Допустимо использовать вложенные циклы.
str_1 = "abc"
str_2 = "d"
str_3 = "efg"
sets = [i+j+k for i in str_1 for j in str_2 for k in str_3]
print(sets)
```

```python
my_tuple = (2, 4, 6)
new_obj = (el+10 for el in my_tuple)

print(new_obj)
```

Здесь мы используем генераторное выражение для элементов кортежа, но в результате получаем объект-итератор. Такой результат связан с использованием круглых скобок в генераторном выражении. Если в этом примере заменить кортеж на список, результат будет идентичный (объект-итератор).

```python
my_tuple = (2, 4, 6)
new_obj = [el+10 for el in my_tuple]

print(new_obj)
```

# ***Модуль random как генератор псевдослучайных чисел***

Модуль random в Python используют для генерации псевдослучайных чисел, а также для случайного выбора и перемешивания элементов последовательностей. Он включает функции для вещественных и целых чисел, работы со списками и задания начального состояния генератора через seed().

Псевдослучайность означает, что значения выглядят случайными, но формируются алгоритмом, а не истинно случайным физическим источником. Это хорошо видно по seed(): если задать одно и то же начальное значение, генератор будет воспроизводить одну и ту же последовательность результатов.

```python
import random

print(random.random())   # число от 0.0 до 1.0
print(random.randint(1, 10))  # целое число от 1 до 10
```

random.random() возвращает вещественное число в диапазоне 0.0 <= N < 1.0, то есть 0 может появиться, а 1 не входит в диапазон. random.randint(a, b) возвращает случайное целое число от a до b включительно.

random.randrange(start, stop, step) выбирает целое число из диапазона с учетом шага, при этом stop не включается. random.choice() берет один случайный элемент из последовательности, random.sample() возвращает несколько уникальных элементов, а random.shuffle() перемешивает список на месте.

```python
n = random.randint(1, 50)
print(n)
```

Такой вызов подходит, когда нужно получить одно случайное целое значение в заданных границах. Если нужен диапазон с шагом, удобнее использовать randrange(), например random.randrange(5, 20, 3).

```python
names = ["Миша", "Инна", "Кристина", "Роман"]
print(random.choice(names))        # один случайный элемент
print(random.sample(names, 3))     # два уникальных элемента
```

choice() выбирает один элемент, а sample() формирует выборку без повторного взятия одной и той же позиции. Для перемешивания всего списка используют shuffle().

Функция random.seed(x) задает начальное состояние генератора. При одинаковом значении seed результаты повторяются, поэтому это удобно для тестов, демонстраций и отладки.

```python
random.seed(1)
print(random.random())

random.seed(5)
print(random.random())   # будет то же самое число
```

В примерах с seed(5) одно и то же число генерируется повторно после повторной инициализации тем же значением. Именно поэтому seed() часто используют, когда нужно воспроизводимое поведение программы.
