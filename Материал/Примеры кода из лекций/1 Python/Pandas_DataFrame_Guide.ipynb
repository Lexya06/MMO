# 2. Создание DataFrame

DataFrame можно создать несколькими способами: из словаря, списков, CSV-файла и других источников.


## 2.1 Из словаря

**Синтаксис:** `pd.DataFrame({'col1': [...], 'col2': [...]})`

Ключи словаря становятся именами столбцов, значения — данными столбцов.


````python
# Создание DataFrame из словаря
df_from_dict = pd.DataFrame({
    'Имя': ['Анна', 'Борис', 'Виктор'],
    'Возраст': [28, 35, 42],
    'Зарплата': [75000, 62000, 95000]
})
print('DataFrame из словаря:')
df_from_dict
````

## 2.2 Из списка списков

**Синтаксис:** `pd.DataFrame([[1,2],[3,4]], columns=['a','b'])`

Каждый вложенный список — одна строка DataFrame.


````python
# Создание DataFrame из списка списков
df_from_list = pd.DataFrame(
    [['Анна', 28, 75000],
     ['Борис', 35, 62000],
     ['Виктор', 42, 95000]],
    columns=['Имя', 'Возраст', 'Зарплата']
)
print('DataFrame из списка списков:')
df_from_list
````

## 2.3 Из списка словарей

**Синтаксис:** `pd.DataFrame([{'a':1,'b':2}, {'a':3,'b':4}])`

Каждый словарь — одна строка; ключи становятся столбцами.


````python
# Создание DataFrame из списка словарей
df_from_records = pd.DataFrame([
    {'Имя': 'Анна', 'Отдел': 'IT', 'Зарплата': 75000},
    {'Имя': 'Борис', 'Отдел': 'Продажи', 'Зарплата': 62000},
    {'Имя': 'Виктор', 'Отдел': 'IT', 'Зарплата': 95000}
])
print('DataFrame из списка словарей:')
df_from_records
````

## 2.4 С указанием индекса

**Синтаксис:** `pd.DataFrame(data, index=[...])`

Параметр `index` позволяет задать пользовательские метки строк.


````python
# Создание DataFrame с пользовательским индексом
df_with_index = pd.DataFrame(
    {'Отдел': ['IT', 'Продажи', 'HR'],
     'Бюджет': [500000, 300000, 150000]},
    index=['dept_1', 'dept_2', 'dept_3']  # пользовательский индекс
)
print('DataFrame с пользовательским индексом:')
df_with_index
````

# 3. Чтение и запись данных

pandas поддерживает чтение и запись данных в различные форматы: CSV, Excel, JSON и другие.


## 3.1 Чтение и запись CSV

**`pd.read_csv(filepath)`** — читает CSV-файл в DataFrame.
**`df.to_csv(filepath)`** — сохраняет DataFrame в CSV.

Основные параметры `read_csv`:
- `sep` — разделитель (по умолчанию `,`)
- `encoding` — кодировка файла
- `header` — номер строки с заголовками
- `index_col` — столбец для использования как индекс

Основные параметры `to_csv`:
- `index=False` — не сохранять индекс
- `sep` — разделитель
- `encoding` — кодировка


````python
# Сохранение основного DataFrame в CSV
df.to_csv('employees.csv', index=False, encoding='utf-8')
print('Файл employees.csv сохранён')

# Чтение CSV обратно
df_loaded = pd.read_csv('employees.csv', encoding='utf-8')
print('\nЧтение из CSV:')
print(df_loaded.head())

# Сохранение с разделителем «точка с запятой» (формат Excel для России)
df.to_csv('employees_semicolon.csv', index=False, sep=';', encoding='utf-8-sig')
print('\nФайл employees_semicolon.csv сохранён (sep=;, utf-8-sig для Excel)')
````

# 5. Выбор данных (индексация)

pandas предоставляет гибкие способы доступа к данным по меткам (`loc`) и позициям (`iloc`).


## 5.1 Выбор столбцов

**`df['col']`** — возвращает столбец как Series.

**`df.col`** — атрибутный доступ (работает если нет пробелов в имени).

**`df[['col1', 'col2']]`** — выбор нескольких столбцов (возвращает DataFrame).


````python
# Выбор одного столбца
print('Столбец Имя (Series):')
print(df['Имя'])

# Выбор нескольких столбцов
print('\nВыбор двух столбцов (DataFrame):')
print(df[['Имя', 'Зарплата']])
````

## 5.2 Выбор строк: loc и iloc

**`df.loc[label]`** — доступ по метке индекса (label-based).

**`df.iloc[pos]`** — доступ по числовой позиции (position-based).

Синтаксис срезов:
- `df.loc[start:end]` — включает обе границы
- `df.iloc[start:end]` — не включает правую границу


````python
# Доступ по метке индекса (loc)
print('Строка с индексом 0 (loc):')
print(df.loc[0])

# Срез строк по меткам
print('\nСтроки 1-3 (loc):')
print(df.loc[1:3])
````

````python
# Доступ по числовой позиции (iloc)
print('Строка на позиции 0 (iloc):')
print(df.iloc[0])

# Срез строк по позиции
print('\nСтроки 0-2 (iloc):')
print(df.iloc[0:3])

# Последняя строка
print('\nПоследняя строка (iloc[-1]):')
print(df.iloc[-1])
````

## 5.3 Выбор ячейки: loc, iloc, at, iat

**`df.loc[row, col]`** — ячейка по метке строки и имени столбца.

**`df.iloc[row_idx, col_idx]`** — ячейка по числовым позициям.

**`df.at[row, col]`** — быстрый скалярный доступ по меткам.

**`df.iat[row_idx, col_idx]`** — быстрый скалярный доступ по позициям.


````python
# Выбор конкретной ячейки через loc
print('df.loc[0, "Имя"]:', df.loc[0, 'Имя'])

# Выбор ячейки через iloc
print('df.iloc[0, 0]:', df.iloc[0, 0])

# Быстрый доступ через at
print('df.at[2, "Зарплата"]:', df.at[2, 'Зарплата'])

# Быстрый доступ через iat
print('df.iat[2, 3]:', df.iat[2, 3])

# Выбор строк и столбцов одновременно
print('\nСтроки 0-2, столбцы Имя и Зарплата:')
print(df.loc[0:2, ['Имя', 'Зарплата']])
````

# 6. Фильтрация данных

Фильтрация позволяет выбирать строки, удовлетворяющие заданным условиям.


## 6.1 Фильтрация по условию

**Синтаксис:** `df[df['col'] > значение]`

Операторы сравнения: `>`, `<`, `>=`, `<=`, `==`, `!=`

Логические операторы:
- `&` — И (AND)
- `|` — ИЛИ (OR)
- `~` — НЕ (NOT)

> Каждое условие при использовании `&` или `|` нужно оборачивать в скобки!


````python
# Сотрудники с зарплатой выше 70 000
print('Зарплата > 70 000:')
print(df[df['Зарплата'] > 70000])

# Несколько условий (И)
print('\nОтдел IT И возраст < 40:')
print(df[(df['Отдел'] == 'IT') & (df['Возраст'] < 40)])

# Несколько условий (ИЛИ)
print('\nОтдел IT ИЛИ зарплата > 80 000:')
print(df[(df['Отдел'] == 'IT') | (df['Зарплата'] > 80000)])
````

## 6.2 query(), isin(), between(), str.contains()

**`df.query('выражение')`** — фильтрация строковым запросом.

**`df['col'].isin([...])`** — проверка вхождения значения в список.

**`df['col'].between(a, b)`** — значения в диапазоне от `a` до `b` включительно.

**`df['col'].str.contains('паттерн')`** — поиск подстроки в строковом столбце.


````python
# Фильтрация через query
print('query: Зарплата > 70000 и Отдел IT:')
print(df.query('Зарплата > 70000 and Отдел == "IT"'))

# isin — проверка вхождения в список
print('\nisin: Отдел — IT или HR:')
print(df[df['Отдел'].isin(['IT', 'HR'])])

# between — диапазон значений
print('\nbetween: Возраст от 30 до 38:')
print(df[df['Возраст'].between(30, 38)])

# str.contains — поиск подстроки
print('\nstr.contains: Имя содержит букву "и":')
print(df[df['Имя'].str.contains('и', case=False, na=False)])
````

# 7. Изменение данных

Добавление, удаление и изменение столбцов и строк DataFrame.


## 7.1 Добавление и удаление столбцов

**`df['new_col'] = values`** — добавление нового столбца.

**`df.drop(columns=['col'])`** — удаление столбцов.

**`df.drop('col', axis=1)`** — альтернативный синтаксис.

**`df.assign(new_col=values)`** — создаёт новый DataFrame с дополнительным столбцом.


````python
# Создаём копию для изменений
df2 = df.copy()

# Добавление нового столбца
df2['Премия'] = df2['Зарплата'] * 0.1
print('После добавления столбца Премия:')
print(df2[['Имя', 'Зарплата', 'Премия']])

# Удаление столбца
df3 = df2.drop(columns=['Премия'])
print('\nПосле удаления столбца Премия:')
print(df3.columns.tolist())

# assign — создаёт новый DataFrame без изменения исходного
df_assigned = df.assign(
    Премия=df['Зарплата'] * 0.1,
    Итого=lambda x: x['Зарплата'] + x['Зарплата'] * 0.1
)
print('\nassign — добавление нескольких столбцов:')
print(df_assigned[['Имя', 'Зарплата', 'Премия', 'Итого']])
````

## 7.2 Удаление строк и переименование столбцов

**`df.drop(index=[0, 1])`** — удаление строк по метке индекса.

**`df.rename(columns={'old': 'new'})`** — переименование столбцов.


````python
# Удаление строк по индексу
df_no_rows = df.drop(index=[0, 1])
print('После удаления строк 0 и 1:')
print(df_no_rows[['Имя', 'Отдел']])

# Переименование столбцов
df_renamed = df.rename(columns={
    'Имя': 'Сотрудник',
    'Зарплата': 'Оклад',
    'Дата_приёма': 'ДатаПриёма'
})
print('\nПосле переименования столбцов:')
print(df_renamed.columns.tolist())
````

## 7.3 Изменение типов и замена значений

**`df['col'].astype(type)`** — преобразование типа данных столбца.

**`df['col'].replace(old, new)`** — замена значений.


````python
# Изменение типа столбца
df_typed = df.copy()
df_typed['Возраст'] = df_typed['Возраст'].astype(float)
print('Тип Возраст после astype(float):', df_typed['Возраст'].dtype)

# Замена значений
df_replaced = df.copy()
df_replaced['Отдел'] = df_replaced['Отдел'].replace({'IT': 'Информационные технологии', 'HR': 'Кадры'})
print('\nПосле замены значений в столбце Отдел:')
print(df_replaced['Отдел'])
````

# 8. Работа с пропущенными значениями (NaN)

Пропущенные значения (`NaN`, `None`) — распространённая проблема реальных данных.
pandas предоставляет удобные инструменты для их поиска, удаления и заполнения.


## 8.1 Обнаружение пропусков

**`df.isna()`** / **`df.isnull()`** — маска True/False для пропущенных значений.

**`df.notna()`** — обратная маска.

**`df.isna().sum()`** — количество пропусков по каждому столбцу.


````python
print('Маска пропущенных значений:')
print(df.isna())

print('\nКоличество пропусков по столбцам:')
print(df.isna().sum())

print('\nОбщее количество пропусков:', df.isna().sum().sum())
````

## 8.2 Удаление строк с пропусками

**`df.dropna()`** — удаляет строки, содержащие хотя бы один NaN.

**`df.dropna(subset=['col'])`** — удаляет строки с NaN только в указанных столбцах.

**`df.dropna(thresh=n)`** — оставляет строки с минимум `n` непустыми значениями.


````python
# Удаление строк с любым NaN
df_no_nan = df.dropna()
print('После dropna() — строк осталось:', len(df_no_nan))
print(df_no_nan[['Имя', 'Рейтинг']])

# Удаление строк с NaN только в столбце Рейтинг
df_no_rating_nan = df.dropna(subset=['Рейтинг'])
print('\nПосле dropna(subset=["Рейтинг"]) — строк осталось:', len(df_no_rating_nan))
print(df_no_rating_nan[['Имя', 'Рейтинг']])
````

## 8.3 Заполнение пропусков

**`df.fillna(value)`** — заполнение константой.

**`df.fillna(method='ffill')`** — заполнение предыдущим значением (forward fill).

**`df.fillna(method='bfill')`** — заполнение следующим значением (backward fill).

**`df.interpolate()`** — интерполяция числовых значений.


````python
# Заполнение NaN средним значением
mean_rating = df['Рейтинг'].mean()
df_filled = df.copy()
df_filled['Рейтинг'] = df_filled['Рейтинг'].fillna(mean_rating)
print('После fillna(mean):')
print(df_filled[['Имя', 'Рейтинг']])

# Forward fill (заполнение предыдущим значением)
df_ffill = df.copy()
df_ffill['Рейтинг'] = df_ffill['Рейтинг'].ffill()
print('\nПосле ffill:')
print(df_ffill[['Имя', 'Рейтинг']])

# Интерполяция
df_interp = df.copy()
df_interp['Рейтинг'] = df_interp['Рейтинг'].interpolate()
print('\nПосле interpolate:')
print(df_interp[['Имя', 'Рейтинг']])
````

# 9. Сортировка

pandas позволяет сортировать DataFrame по значениям столбцов или по индексу.


## 9.1 sort_values()

**`df.sort_values('col')`** — сортировка по возрастанию.

**`df.sort_values('col', ascending=False)`** — по убыванию.

**`df.sort_values(['col1', 'col2'])`** — по нескольким столбцам.


Параметр `na_position`: `'last'` (NaN в конец) или `'first'` (NaN в начало).

# 11. Объединение DataFrame

pandas предоставляет несколько способов объединения данных: конкатенацию и слияние (аналог SQL JOIN).


## 11.1 concat() — вертикальное и горизонтальное объединение

**`pd.concat([df1, df2])`** — вертикальное объединение (добавление строк).

**`pd.concat([df1, df2], axis=1)`** — горизонтальное объединение (добавление столбцов).

Параметр `ignore_index=True` сбрасывает индекс в результирующем DataFrame.


````python
# Создаём дополнительный DataFrame новых сотрудников
new_employees = pd.DataFrame({
    'Имя': ['Игорь', 'Карина'],
    'Отдел': ['Маркетинг', 'HR'],
    'Возраст': [27, 33],
    'Зарплата': [55000, 61000],
    'Дата_приёма': pd.to_datetime(['2023-01-10', '2023-03-01']),
    'Рейтинг': [4.0, 3.9]
})

# Вертикальное объединение
df_concat = pd.concat([df, new_employees], ignore_index=True)
print('После вертикального объединения (concat):')
print(df_concat[['Имя', 'Отдел', 'Зарплата']])
````

## 11.2 merge() — слияние по ключу

**Синтаксис:** `pd.merge(df1, df2, on='key', how='inner')`

Типы слияния (`how`):
- `'inner'` — только совпадающие строки (по умолчанию)
- `'left'` — все строки из левого + совпадения из правого
- `'right'` — все строки из правого + совпадения из левого
- `'outer'` — все строки из обоих DataFrame


````python
# Слияние сотрудников с информацией об отделах
print('departments:')
print(departments)

print('\nСотрудники:')
print(df)

# Inner join (только совпадающие отделы)
df_inner = pd.merge(df, departments, on='Отдел', how='inner')
print('\n\nInner merge (сотрудники + отделы):')
print(df_inner[['Имя', 'Отдел', 'Зарплата', 'Бюджет', 'Руководитель']])

# Left join (все отделы, NaN если сотрудник не найден)
df_right = pd.merge(df, departments, on='Отдел', how='right')
print('\nRight merge:')
print(df_left[['Имя', 'Отдел', 'Зарплата', 'Бюджет', 'Руководитель']])
````

# 12. Применение функций

pandas позволяет применять функции к столбцам, строкам и отдельным элементам.


## 12.1 apply() для столбцов и строк

**`df['col'].apply(func)`** — применяет функцию к каждому элементу столбца.

**`df.apply(func, axis=0)`** — применяет функцию к каждому столбцу.

**`df.apply(func, axis=1)`** — применяет функцию к каждой строке.


````python
# apply к столбцу
def classify_salary(salary):
    if salary >= 80000:
        return 'Высокая'
    elif salary >= 65000:
        return 'Средняя'
    else:
        return 'Низкая'

df_apply = df.copy()
df_apply['Категория_зп'] = df_apply['Зарплата'].apply(classify_salary)
print('Классификация зарплаты через apply:')
print(df_apply[['Имя', 'Зарплата', 'Категория_зп']])

# apply к строке (axis=1)
df_apply['Сводка'] = df_apply.apply(
    lambda row: f"{row['Имя']} ({row['Отдел']}, {row['Зарплата']} руб.)",
    axis=1
)
print('\nСводка по строкам (axis=1):')
print(df_apply['Сводка'])
````

## 12.2 apply() к столбцам (axis=0) и лямбда-функции

**`df.apply(func, axis=0)`** — применяет функцию к каждому столбцу целиком (получает Series).

**Лямбда-функции** — краткий синтаксис для простых преобразований: `lambda x: выражение`


````python
# apply к каждому столбцу (axis=0)
numeric_cols = df[['Возраст', 'Зарплата', 'Рейтинг']].copy()
print('Минимум и максимум по каждому столбцу:')
print(numeric_cols.apply(lambda col: col.max() - col.min(), axis=0))

# Лямбда-функции
df_lambda = df.copy()
df_lambda['Зарплата_тыс'] = df_lambda['Зарплата'].apply(lambda x: round(x / 1000, 1))
print('\nЗарплата в тысячах рублей:')
print(df_lambda[['Имя', 'Зарплата', 'Зарплата_тыс']])
````

## 12.3 map() — поэлементное применение

**`df['col'].map(dict_or_func)`** — заменяет значения столбца, используя словарь или функцию.

**`df.map(func)`** (или `df.applymap(func)` в старых версиях) — применяет функцию к каждой ячейке.


````python
# map со словарём — замена значений
dept_map = {'IT': 'Технологии', 'Продажи': 'Sales', 'HR': 'Кадры'}
df_mapped = df.copy()
df_mapped['Отдел_EN'] = df_mapped['Отдел'].map(dept_map)
print('map со словарём:')
print(df_mapped[['Имя', 'Отдел', 'Отдел_EN']])

# DataFrame.map — к каждой ячейке
numeric_df = df[['Возраст', 'Зарплата']].copy()
result = numeric_df.map(lambda x: x * 1.1 if pd.notna(x) else x)
print('\nДанные увеличены на 10% (df.map):')
print(result)
````

## 14.1 Преобразование в datetime и извлечение компонентов

````python
# Текущая дата
today = pd.Timestamp('2026-01-01')  # фиксированная дата для воспроизводимости

# Вычисление стажа
df_dates['Стаж_дней'] = (today - df_dates['Дата_приёма']).dt.days
df_dates['Стаж_лет'] = (df_dates['Стаж_дней'] / 365.25).round(1)

print('Стаж сотрудников:')
print(df_dates[['Имя', 'Дата_приёма', 'Стаж_дней', 'Стаж_лет']])
````

````python
# Добавление временного интервала
df_dates['Дата_пробации'] = df_dates['Дата_приёма'] + pd.Timedelta(days=90)
print('\nДата окончания испытательного срока:')
print(df_dates[['Имя', 'Дата_приёма', 'Дата_пробации']])
````

**`df.drop_duplicates(subset=['col'])`** — проверяет дубли только по указанным столбцам.


Параметр `keep`: `'first'` (оставить первое), `'last'` (оставить последнее), `False` (удалить все).


````python
# Создаём DataFrame с дубликатами
df_with_dupes = pd.concat([df, df.iloc[[0, 2]]], ignore_index=True)
print('DataFrame с дубликатами:')
print(df_with_dupes[['Имя', 'Отдел', 'Зарплата']])
````

````python
# Маска дубликатов
print('\nМаска дубликатов (duplicated):')
print(df_with_dupes.duplicated())
````

````python
# Количество дубликатов
print('\nКоличество дубликатов:', df_with_dupes.duplicated().sum())
````

````python
# Удаление всех дубликатов
df_no_dupes = df_with_dupes.drop_duplicates()
print('После drop_duplicates():')
print(df_no_dupes[['Имя', 'Отдел', 'Зарплата']])
print('Строк после удаления дубликатов:', len(df_no_dupes))
````

````python
# Удаление дубликатов по конкретному столбцу
df_unique_dept = df_with_dupes.drop_duplicates(subset=['Отдел'])
print('\nУдаление по столбцу Отдел (первый сотрудник из каждого отдела):')
print(df_unique_dept[['Имя', 'Отдел']])
````

# 18. Настройки отображения

pandas позволяет настраивать формат вывода DataFrame в Jupyter/Colab.


## 18.1 pd.set_option() и pd.reset_option()

Основные опции:
- **`display.max_rows`** — максимальное количество отображаемых строк
- **`display.max_columns`** — максимальное количество отображаемых столбцов
- **`display.max_colwidth`** — максимальная ширина столбца (символов)
- **`display.float_format`** — формат отображения чисел с плавающей точкой
- **`display.precision`** — число знаков после запятой

**`pd.reset_option('all')`** — сброс всех настроек к значениям по умолчанию.


````python
# Текущие настройки
print('Текущие значения:')
print('max_rows:', pd.get_option('display.max_rows'))
print('max_columns:', pd.get_option('display.max_columns'))
print('max_colwidth:', pd.get_option('display.max_colwidth'))
````

````python
# Установка новых настроек
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_colwidth', 200)
pd.set_option('display.float_format', '{:.2f}'.format)

print('\nПосле настройки float_format — числа с 2 знаками после запятой:')
print(df[['Имя', 'Зарплата', 'Рейтинг']])
````

````python
# Установка числа знаков после запятой
pd.set_option('display.precision', 3)
print('Точность 3 знака:')
print(df[['Имя', 'Рейтинг']])
````

````python
# Сброс всех настроек
pd.reset_option('all')
print('\nНастройки сброшены до значений по умолчанию')
print('max_rows после сброса:', pd.get_option('display.max_rows'))
````
