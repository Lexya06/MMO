# ***Понятие класса***

Класс в Python — это шаблон, по которому создаются объекты с общими свойствами и поведением. В классе обычно описывают данные объекта через атрибуты и его действия через методы, а сам объект является экземпляром этого класса.

Класс нужен, чтобы объединить связанные данные и операции над ними в одной структуре. Такой подход делает код более понятным, особенно когда в программе есть сущности вроде студента, книги, автомобиля или пользователя.



Например, если нужно описать студента, удобно задать его имя, группу и средний балл как атрибуты, а вывод информации или расчет чего-либо — как методы. То есть класс задает общую модель, а каждый объект хранит уже свои конкретные значения.

```python
class Student:
    def __init__(self, name, group):
        self.name = name
        self.group = group

    def show_info(self):
        print("Студент:", self.name, "Группа:", self.group)
```

Здесь Student — это класс, __init__() — специальный метод инициализации объекта, self.name и self.group — атрибуты экземпляра, а show_info() — метод класса. Метод self использует ссылку на конкретный объект, для которого он вызван.

Создание объекта выглядит так:

```python
s1 = Student("Ольга", "107001")
s1.show_info()
```

Переменная s1 хранит экземпляр класса Student, созданный по описанию класса. Один и тот же класс позволяет создавать много объектов с одинаковой структурой, но разными значениями атрибутов.

У класса обычно есть:


*   **Атрибуты** — данные объекта, например имя, возраст, номер группы.

*   **Методы** — функции внутри класса, которые описывают поведение объекта.

*   Конструктор **__init__**() — специальный метод, который срабатывает при создании экземпляра и помогает задать начальные значения.

Важно различать класс и объект: класс — это описание, а объект — конкретный экземпляр по этому описанию. Часто это объясняют как чертеж и построенный по нему дом.

**self** в Python — это ссылка на текущий экземпляр класса, то есть на тот объект, с которым сейчас работает метод. Обычно self указывают первым параметром метода экземпляра, чтобы внутри метода обращаться к атрибутам и другим методам именно этого объекта.

Через self метод получает доступ к данным конкретного объекта, например к self.name или self.age. Это позволяет разным объектам одного класса хранить свои собственные значения, а не использовать одну общую переменную для всех. Кроме того, self помогает отличать атрибуты объекта от обычных локальных переменных внутри метода.


```python
class Student:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)

s1 = Student("Ольга")
s1.show_name()
```

self не является ключевым словом Python, это просто общепринятое имя первого параметра метода экземпляра.

Теоретически его можно назвать иначе, но в реальном коде почти всегда пишут именно self, потому что это стандарт Python и так код легче читать.

**self** — это ссылка метода на сам объект.

__init__ в Python — это специальный метод класса, который автоматически вызывается сразу после создания нового объекта и обычно используется для задания начальных значений его атрибутов. Его часто называют конструктором, хотя точнее сказать, что создание объекта связано с __new__, а __init__ отвечает именно за его инициализацию.

Метод __init__ объявляют внутри класса, а первым параметром по соглашению указывают self, который ссылается на текущий экземпляр объекта. Когда вы пишете obj = MyClass(...), Python создает объект и затем передает в __init__ сам объект и аргументы, которые были указаны при создании.



```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Внутри __init__ обычно сохраняют переданные значения в атрибуты объекта, например self.name = name.

Это позволяет каждому экземпляру класса иметь собственные данные, заданные в момент создания.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Ольга", 20)
print(student1.name)   # Ольга
print(student1.age)    # 20
```

В примере выше Student — это класс, а student1 — его экземпляр.

При вызове Student("Ольга", 20) метод __init__ получает "Ольга" и 20, после чего записывает их в атрибуты self.name и self.age.

Благодаря этому у объекта сразу есть готовое начальное состояние.
Метод __init__ ничего не должен возвращать; его задача — настроить объект, а не создать новый результат.
​
У метода могут быть параметры по умолчанию, если вы хотите разрешить создание объектов без обязательной передачи всех значений.





```python
class User:
    def __init__(self, name="Гость"):
        self.name = name

u1 = User()
u2 = User("Анна")

print(u1.name)   # Гость
print(u2.name)   # Анна
```

__init__ — это специальный метод класса в Python, который автоматически вызывается при создании объекта и инициализирует его атрибуты.

# ***Модификаторы доступа***

Модификаторы доступа — это способ показать, какие атрибуты и методы класса предназначены для внешнего использования, а какие считаются внутренними. В Python нет жесткой системы модификаторов доступа, как в Java или C++, поэтому здесь используются соглашения об именовании. В Python обычно выделяют три уровня доступа: public, protected и private. Public-элементы доступны отовсюду, protected считаются внутренними для класса и его наследников, а private предназначены только для использования внутри самого класса.

Обозначаются они так:

name — публичный атрибут или метод.

_name — защищенный атрибут или метод по соглашению.

__name — приватный атрибут или метод с механизмом переименования имени.

```python
# Приватный метод (свойство) (он же private) - такой метод (свойство),
# доступ к которому можно получить только из того же класса (или объекта того же класса).

# Защищенный метод (свойство) (он же protected) - такой метод (свойство),
# доступ к которому можно получить только из того же класса (или объекта того же класса) и из его наследников.

# Публичный метод (свойство) (он же public) - такой метод (свойство),
# доступ к которому можно получить откуда угодно - извне определенного класса,
# из объекта определенного класса, из наследников.
```

Публичные атрибуты и методы не имеют подчеркивания в начале имени и предназначены для обычного использования извне класса. Их можно свободно читать и изменять через объект.

```python
class Student:
    def __init__(self, name):
        self.name = name   # public

s = Student("Ольга")
print(s.name)
```

Защищенные элементы записываются с одним подчеркиванием, например _balance, и это сигнал программисту, что напрямую снаружи их лучше не использовать без необходимости. Такой доступ не запрещен технически, но считается внутренним соглашением Python.



```python
class Account:
    def __init__(self, money):
        self._money = money   # protected
```

Приватные элементы записываются с двумя подчеркиваниями, например __password, и Python применяет к ним name mangling — изменяет имя так, чтобы уменьшить риск случайного доступа или переопределения. Это не абсолютная защита, но такой механизм делает обращение снаружи менее прямым.

```python
class User:
    def __init__(self, password):
        self.__password = password   # private

    def show(self):
        print(self.__password)
```

Главное, что нужно запомнить: в Python модификаторы доступа — это в первую очередь соглашения и механизм инкапсуляции, а не строгий запрет на доступ. Даже приватный атрибут при желании можно получить через измененное имя вида _ИмяКласса__атрибут, хотя в обычной практике так делать не рекомендуется.

# ***Наследование***

Наследование в Python — это механизм ООП, при котором новый класс создается на основе уже существующего и получает его атрибуты и методы. Исходный класс называют базовым или родительским, а новый — дочерним или производным. Наследование нужно, чтобы не дублировать код и повторно использовать уже описанное поведение в новых классах.  Дочерний класс может не только получить возможности родителя, но и добавить свои собственные атрибуты и методы или изменить унаследованные.



```python
class Animal:
    def speak(self):
        print("Какой-то звук")

class Dog(Animal):
    pass
```

Здесь Dog наследуется от Animal, поэтому объект класса Dog может использовать метод speak(), даже если он не описан внутри Dog

Когда один родительский класс имеет несколько дочерних классов, каждый потомок наследует общие свойства базового класса и при этом может добавлять свои методы или переопределять существующие.

Так строят иерархию, где общий код выносится в родителя, а различия описываются в потомках.



```python
class Animal:
    def eat(self):
        print("Есть")

class Dog(Animal):
    def speak(self):
        print("Гав")

class Cat(Animal):
    def speak(self):
        print("Мяу")
```

Здесь и Dog, и Cat наследуют общий метод eat() от Animal, но имеют собственную реализацию speak().

При множественном наследовании потомок получает атрибуты и методы всех указанных родителей. Синтаксис такой: после имени дочернего класса в скобках перечисляют несколько базовых классов через запятую.

```python
class Car:
    def move(self):
        print("Едет по дороге")

class Boat:
    def swim(self):
        print("Плывет по воде")

class AmphibiousCar(Car, Boat):
    pass

a = AmphibiousCar()
a.move()
a.swim()
```

В этом примере AmphibiousCar наследует поведение сразу от двух родителей: Car и Boat.

Если у родителей есть методы с одинаковыми именами, Python выбирает их по порядку разрешения методов, который называется MRO. Этот порядок зависит от того, как родители перечислены в объявлении класса, и его можно посмотреть через mro() или __mro__; при работе с __init__ в таких схемах обычно используют super()

```python
class A:
    def show(self):
        print("Метод из A")

class B:
    def show(self):
        print("Метод из B")

class C(A, B):
    pass

obj = C()
obj.show()          # Метод из A
print(C.mro())
```

В записи class C(A, B) метод show() будет найден сначала в A, а затем уже в B, если в A его нет.

Вот простой пример множественного наследования, где у двух родителей есть одинаковый метод show().

```python
class A:
    def show(self):
        print("Метод класса A")

class B:
    def show(self):
        print("Метод класса B")

class C(A, B):
    pass

obj = C()
obj.show()
```

Дочерний класс может переопределить метод родителя, то есть написать свою версию метода с тем же именем. Это позволяет сохранить общую структуру, но изменить поведение для конкретного класса.

# ***Переопределение методов***

```python
class Animal:
    def speak(self):
        print("Какой-то звук")

class Dog(Animal):
    def speak(self):
        print("Гав-гав")
```

В этом примере Dog наследует класс Animal, но использует уже собственную реализацию метода speak()

Если дочерний класс определяет свой __init__, то для использования инициализации родителя обычно вызывают super().__init__(...).

Это помогает сначала задать общие свойства из базового класса, а потом добавить поля самого дочернего класса.


```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model
```

Здесь Car получает атрибут brand от Vehicle, а затем добавляет собственный атрибут model.

# ***Интерфейсы***

```python
from abc import ABC, abstractmethod

class MyAbstractClass(ABC):
    @abstractmethod
    def my_method_1(self):
        pass

    @abstractmethod
    def my_method_2(self):
        pass

class MyClass(MyAbstractClass):
    pass
```

```python
mc = MyClass()
```

В этом примере создается абстрактный класс MyAbstractClass и в случае наследования от него во всех классах-потомках необходимо реализовать два базовых метода, т. е., все классы-потомки наследуют интерфейс родителя. Соответственно, логику класса MyClass в примере выше необходимо изменить





```python
from abc import ABC, abstractmethod

class MyAbstractClass(ABC):
    @abstractmethod
    def my_method_1(self):
        print("Метод my_method_0()")
        pass
    @abstractmethod
    def my_method_2(self):
        print("Метод my_method_00()")
        pass

class MyClass(MyAbstractClass):
    def my_method_1(self):
        print("Метод my_method_1()")

    def my_method_2(self):
        print("Метод my_method_2()")
```

```python
mc = MyClass()
mc.my_method_1()
```

# ***Декоратор @property***

@**property** в Python — это декоратор, который превращает метод класса в свойство, поэтому к нему можно обращаться как к обычному атрибуту, без круглых скобок. Его используют, когда нужно скрыть внутренний атрибут, добавить проверку значения или оставить удобный интерфейс вида obj.value вместо obj.get_value().

Метод с @property становится геттером, то есть отвечает за чтение значения свойства. Если нужно разрешить изменение значения, рядом объявляют метод с декоратором @имя_свойства.setter, а для удаления можно использовать @имя_свойства.deleter.

```python
class Product:
    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Цена не может быть отрицательной")
        self._price = value
```

В этом примере снаружи используется product.price, но внутри класса значение хранится в _price, и при присваивании автоматически выполняется проверка.

```python
p = Product(100)
print(p.price)   # 100
p.price = 250
print(p.price)   # 250
```

@property помогает контролировать доступ к данным объекта, не ломая привычный синтаксис работы с атрибутами.
Это особенно полезно, если позже вы хотите добавить проверку, вычисление значения или защиту атрибута, но при этом сохранить тот же внешний интерфейс класса.
